## **EV Charging Infrastructure Analysis**

*Notebook 02 — Data Audit*

This notebook performs a detailed audit of the dataset to identify potential data quality issues before the cleaning stage.

## Objectives

- Inspect the dataset structure.
- Identify missing values.
- Identify duplicated records.
- Analyze data types.
- Check unique values in categorical variables.
- Identify potential inconsistencies.
- Analyze numerical variables and their distributions.
- Identify potential outliers.

## 1. Libraries and Dataset

```python
# Import libraries
import pandas as pd
import numpy as np
```

## 2. Load Dataset

The dataset will be loaded using Pandas' `read_csv()` function.

```python
# Define dataset path
DATA_PATH = "../data/raw/global_ev_market_charging_infrastructure_2026.csv"
```

```python
# Load dataset
df = pd.read_csv(DATA_PATH)
```

## 3. Missing Values

Missing values can affect data analysis and Machine Learning models. Therefore, they must be identified before the cleaning stage.

```python
# Missing Values

missing_values = df.isnull().sum()
display(missing_values)

# Missing value percentage
missing_percentage = (missing_values/len(df)) * 100
display(missing_percentage)

# Missing values summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
    }).sort_values(by="Missing Percentage", ascending=False)

display(missing_summary)
```

## 4. Duplicate Records

Duplicate records can lead to biased analysis and may affect Machine Learning models. We will check whether the dataset contains completely duplicated rows.

```python
# Duplicated rows
duplicated_rows = df.duplicated().sum()
print(f"Number of duplicated rows: {duplicated_rows}")
```

## 5. Record IDs

The `record_id` column will be analyzed to determine whether each record has a unique identifier.

```python
# Unique record IDs
unique_ids = df["record_id"].nunique()
print(f"Number of records: {len(df):,}")
print(f"Number of unique IDs: {unique_ids:,}")
```

```python
# Duplicated record IDs
duplicated_ids = df["record_id"].duplicated().sum()
print(f"Number of duplicated IDs: {duplicated_ids}")
```

## 6. Categorical Variables

Categorical variables will be inspected to identify their unique values and detect potential inconsistencies.

```python
categorical_columns = df.select_dtypes(include=object).columns
print("Categorical columns:")
print(categorical_columns)
```

```python
# Country values
display(df["country"].value_counts())
```

```python
# Region values
display(df["region"].value_counts())
```

```python
# Vehicle type values
display(df["vehicle_type"].value_counts())
```

```python
# Government incentive categories
display(df["govt_incentives"].value_counts())
```

## 7. Numerical Variables

Numerical variables will be analyzed using descriptive statistics to identify their ranges, central tendencies, and potential anomalies.

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(numerical_columns)
```

```python
# Descriptive statistics
display(df[numerical_columns].describe().T)
```

## 8. Unique Values

The number of unique values in each numerical variable will be inspected to identify potential inconsistencies or unexpected values.

```python
# Number of unique values
unique_values = df[numerical_columns].nunique()

display(unique_values)
```

## 9. Numerical Ranges

Minimum and maximum values will be inspected to identify potentially invalid or unrealistic values.

```python
# Minimum values
display(df[numerical_columns].min())
```

```python
# Minimum values
display(df[numerical_columns].min())
```

## 10. Market Adoption Rate

The market adoption rate will be inspected separately because it represents a rate and should follow a specific range.

```python
# Market adoption rate range
print(f"Minimum: {df['market_adoption_rate'].min()}")
print(f"Maximum: {df['market_adoption_rate'].max()}")
```

## 11. Year Distribution

The distribution of records across years will be inspected to identify the temporal range and potential inconsistencies.

```python
# Year range
print(f"Minimum year: {df['year'].min()}")
print(f"Maximum year: {df['year'].max()}")
```
